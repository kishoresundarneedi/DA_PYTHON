```python
# LOWER CASE, UPPER CASE, CAPITALIZE, TITLE, SWAP CASE, CASE FOLD.
a = "prasanna kumar is very good boy"
```

```python
print(a.lower())
print(a.upper())
print(a.title())
print(a.capitalize())
print(a.swapcase())
print(a.casefold())
```

```python
print(a.swapcase())
print(a.casefold())
```
