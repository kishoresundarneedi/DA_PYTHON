Dictionary in python :
==========================
Each value is associated with a unique key.
Dictionaries are mutable(we can modify them).
Dictionaries are ordered(maintain insertion order).
Keys must be unique.
Values can be dulicated.
Keys should be immutable data types such as strings,integers,or tuples.

```python
student={"name":"kishore","age":"22","marks":"99"}
print(student)
```

```python
print(student["name"])
print(student["age"])
print(student["marks"])
```

```python
print(type(student))
```

```python
# Adding new element.
student["course"]="python"
print(student)
```

```python
student["marks"]="100"
print(student)
```

```python
# Updating Dict
student["age"]="25"
student["marks"]="99"
print(student)
```

```python
# create a dictionary of employee details with 5 details ?
employee={"name":"kishore","salary":"90000","company":"google","role":"softwaretrainer","employee_id":"101"}
print(employee)
```

```python
employee["salary"]="120000"
print(employee)
```

```python
employee["blood group"]="o-"
print(employee)
```

```python
employee.pop("salary")
print(employee)
```

```python
employee.pop("blood group")
print(employee)
```

```python
employee.popitem()
print(employee)
```

```python
employee.clear()
print(employee)
```

```python
for i in student.keys():
  print(i)
```

```python
for i in student.values():
  print(i)
```

```python
for j,k in student.items():
  print(f"{j}:{k}")
```
