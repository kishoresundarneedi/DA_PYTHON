```python
user=input()
pswd=input()
while True:
  u=input("enter username")
  p=input("enter password")
  if user ==u and pswd ==p:
    print("login successful")
    break
  elif user!=u and pswd==p:
   print("incorrect username")
  elif user ==u and pswd!=p:
    print("incorrect password")
  elif user!=u and pswd!=p:
    print("both are incorrect")
  else:
    print("enter correct details")
```

```python
# "atm pin verification"
atm=int(input())
while True:
  if atm<999 or atm>9999:
    print("invalid pin")
    break
  pin=int(input("enter your pin :"))
  if atm==pin:
    print("money credit")
    break
  else:
    print("incorrect pin")
```
